## Bar Chart

```python
plt.bar(cities, population / 1000000)
plt.title("Population of 10 Industrial Cities in Iran")
plt.xlabel("Cities")
plt.ylabel("Population (millions)")
plt.xticks(rotation=90)
plt.ylim(0, 9)
plt.show()

```

## Horizontal Bar Chart

```python
plt.barh(cities, population / 1000000)
plt.title("Population of 10 Industrial Cities in Iran")
plt.xlabel("population (millions)")
plt.ylabel("Cities")
plt.xlim(0, 9)
plt.show()
```

## Line Plot

```python
plt.plot(cities, population / 1000000)
plt.title("Population of 10 Industrial Cities in Iran")
plt.xlabel("Cities")
plt.ylabel("Population (millions)")
plt.xticks(rotation=90)
plt.ylim(0, 9)
plt.show()
```
